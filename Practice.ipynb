```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sns.__version__
```

```python
data = {
    'Hours':[1,2,3,4,5],
    'Marks':[20,40,60,80,100]
}

df =pd.DataFrame(data)

sns.scatterplot(x='Hours',y='Marks',data=df)
plt.show()
```

```python
data = {
    'Ad_Budget':[10,15,20,25,30,35,40],
    'Sales':[50,65,80,95,120,130,150]
}

df = pd.DataFrame(data)

sns.scatterplot(x='Ad_Budget',y='Sales',data= df)

plt.title('Advertising Budget vs Sales')
plt.show()
```

```python
tips = sns.load_dataset('tips')
print(tips.head())

sns.scatterplot(x='total_bill',y='tip',data=tips)

plt.show()
```

```python
flights = sns.load_dataset('flights')

sns.lineplot(x='year',y='passengers',data=flights)

plt.show()
```

```python
tips = sns.load_dataset('tips')
sns.barplot(x='day',y='total_bill',data=tips)

plt.show()
```

```python
sns.histplot(tips['total_bill'])

plt.show()
```

```python
marks = [20,25,30,35,40,45,50,55,60,95]

sns.boxplot(y=marks)

plt.title('Student Marks')
plt.show()
```

```python
salary = [10000,12000,15000,18000,20000,
          22000,25000,28000,30000,100000]
sns.boxplot(y=salary)

plt.title('Employee Salary')
plt.show()
```

```python
tips = sns.load_dataset('tips')

print(tips.head())

sns.boxplot(x='day',y='total_bill',data=tips)

plt.show()
```

```python
sns.countplot(x='day',data=tips)
plt.show()
```

```python
data = [
    [10,20,30],
    [40,50,60],
    [70,80,90]
]

sns.heatmap(data,annot=True,cmap='Greens')
plt.show()
```

```python
data = {
    'Math':[80,70,90],
    'Science':[75,88,92],
    'English':[85,79,91]
}
df = pd.DataFrame(data)
sns.heatmap(df,annot=True,cmap='Reds')
plt.show()
```

```python
sns.heatmap(df,annot=True,cmap='Blues',linewidths=2,linecolor='red',fmt='.0f')
plt.title('Student Marks')
plt.xlabel('Subjects')
plt.ylabel('Students')

plt.show()
```
